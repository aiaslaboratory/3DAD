```python
import numpy as np
from sklearn.metrics import roc_auc_score, accuracy_score,precision_recall_curve, auc
from sklearn.metrics import confusion_matrix
```

```python

def read_scores(file_path):
    scores = []
    with open(file_path, 'r') as file:
        lines = file.readlines()
        # 除去最后一行的所有行
        for line in lines[:-1]:
            try:
                score = float(line.strip())
                scores.append(score)
            except ValueError:
                # 如果转换失败，则跳过这一行
                continue
    return scores
```

```python
file_paths = ['txt/good_anomaly_scores.txt','txt/combined_anomaly_scores.txt', 'txt/cut_anomaly_scores.txt', 'txt/contamination_anomaly_scores.txt']
all_scores = []
all_labels = []

# 读取数据并分配标签
for i, file_path in enumerate(file_paths):
    scores = read_scores(file_path)
    all_scores.extend(scores)
    # 前一个文件是正常数据（标签0），其余是异常（标签1）
    labels = [0 if i == 0 else 1 for _ in scores]
    all_labels.extend(labels)

# 将列表转换为NumPy数组以方便处理
all_scores = np.array(all_scores)
all_labels = np.array(all_labels)

# 设定阈值a


# 应用阈值进行分类


auc_roc = roc_auc_score(all_labels, all_scores)

precision, recall, _ = precision_recall_curve(all_labels, all_scores)

# 计算曲线下面积（AUC）
auc_pr = auc(recall, precision)

print(f"AUC-PR: {auc_pr}")
print(f"AUC-ROC: {auc_roc}")
```

```python
import matplotlib.pyplot as plt

# 定义函数来读取.log文件并提取数据
def read_log(file_path):
    
    epoch = []
    loss = []
    val_loss = []
    
    with open(file_path, 'r') as file:
        next(file)
        for line in file:
            parts = line.strip().split(',')
            if len(parts) == 3:  # 确保每一行都是正确的格式
                epoch.append(int(parts[0]))
                loss.append(float(parts[1]))
                val_loss.append(float(parts[2]))
                
    return epoch, loss, val_loss

# 文件路径，请替换为你的.log文件的实际路径
file_path = 'logs/peach_kdtraining_loss.log'
epochs, losses, val_losses = read_log(file_path)

# 使用matplotlib绘制图像
plt.figure(figsize=(12, 6))

# 绘制训练损失曲线
plt.plot(epochs, losses, label='Training Loss', color='blue')

# 绘制验证损失曲线
plt.plot(epochs, val_losses, label='Validation Loss', color='red', linestyle='--')

# 设置图表标题和坐标轴标签
plt.title('Training and Validation Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

# 显示图像
plt.grid(True)
plt.show()
```
